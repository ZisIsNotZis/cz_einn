````python
from fastai.vision.all import *
from torch.nn import *

torch.cuda.set_device(1)


class Dat:
    a = torch.from_numpy(np.memmap(f"../tmp/news2016zh.H", "h"))

    def __init__(s, w, b, n=0):
        s.a = s.a.as_strided([len(s.a) - w + 1, w], [1, 1])
        s.b = b
        s.n = n

    def __len__(s):
        return s.n or len(s.a) // s.b

    def __iter__(s):
        for x in range(len(s)):
            x = s.a[torch.randint(len(s.a), [s.b])].to("cuda", torch.long)
            yield x, x.flatten()


class F(Module):
    def __init__(s, f):
        super().__init__()
        s.forward = f
        for i, f in zip(f.__code__.co_varnames[1:], f.__defaults__ or ()):
            setattr(s, i, f)
````

````python
m = Learner(
    DataLoaders(Dat(8, 512*5, 99999), Dat(8, 512*5, 1)),
    Sequential(
        Embedding(21482, 8),
        *[F(lambda x: x.view(-1, 16)), Linear(16, 8), CELU(1)] * 3,
        Linear(8, 8),
        *[CELU(1), Linear(8, 16), F(lambda x: x.view(-1, 8))] * 3,
        Linear(8, 21482),
    ).cuda(),
    CrossEntropyLoss(),
    metrics=accuracy,
    cbs=ShowGraphCallback,
)
m.lr_find()
````

````python
m.fit_one_cycle(10, .001, 9, 999, .01)
# m.save('einn5-autoencoder-Copy1')
````
