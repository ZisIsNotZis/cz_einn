```python
from functools import lru_cache
from itertools import zip_longest


def tolist(f):
    def tolist(*a):
        return tuple(f(*a))

    return tolist


@lru_cache(None)
@tolist
def mkindex(d, n):
    if n == 1:
        yield ((0,),)
    else:
        t = 2 ** (d - 1) - 1
        for i in range(max(1, n - min(t, n - 1)), min(t, n - 1) + 1):
            for l in mkindex(d - 1, i):
                for r in mkindex(d - 1, n - i):
                    yield ((),) + tuple(l + tuple(2 ** i + r for r in r) for i, (l, r) in enumerate(zip_longest(l, r, fillvalue=())))


def pakindex(d, n):
    a = mkindex(d, n)
    child = np.fromiter((2 ** i - 1 + a for a in a for i, a in enumerate(a) for a in a), "i").reshape(-1, n)
    parent = []
    t = child
    for _ in range(d - 1):
        t = (t - 1).clip(0) // 2
        parent.append(t)
    parent = np.array([list(set(i)) for i in np.concatenate(parent, 1)], "i")
    return child, parent


from fastai.vision.all import *
from torch.nn import *


class Dat:
    a = torch.from_numpy(np.memmap(f"../tmp/news2016zhMin100.H", "h"))
    m = np.array([open("../tmp/news2016zhMin100.txt").read()]).view("U1")

    def __init__(s, w, b, n=1):
        s.a = s.a.as_strided([len(s.a) - w + 1, w], [1, 1])
        s.b = b
        s.n = n

    def __len__(s):
        return len(s.a) // (s.a.size(1) * s.b * s.n)

    def __iter__(s):
        for x in range(len(s)):
            x = s.a[torch.randint(len(s.a), [s.b])].to("cuda", torch.long)
            yield x, x


class Transpose(Module):
    @staticmethod
    def forward(x):
        return x.transpose(1, 2)


class ConvBn(Sequential):
    def __init__(s, i, o=0):
        o = o or i
        super().__init__(Conv1d(i, o, 3, padding=1, bias=0), BatchNorm1d(o, track_running_stats=0), ReLU(1))


class Res(Sequential):
    def forward(s, x):
        return super().forward(x) + x


class Cnn(Sequential):
    def __init__(s, *a, n=0):
        super().__init__(*a)
        s.n = n

    def forward(s, x):
        for _ in range(s.n or int(math.log2(x.size(2)))):
            x = super().forward(x)
        return x


class View(Module):
    def __init__(s, i):
        super().__init__()
        s.i = i

    def forward(s, x):
        return x.view(len(x), -1, s.i)


class Tnn(Sequential):
    def __init__(s, *a, n=0):
        super().__init__(*a)
        s.n = n

    def forward(s, x):
        y = [x]
        for _ in range(s.n):
            x = super().forward(x)
            y.append(x)
        return torch.cat(y, 1)


class MultiCrossEntropyLoss(Module):
    def forward(s, x, y):
        n = (x.size(1) + 1) // 2
        X, x = x[..., :-1], x[..., -1]
        X = X.log_softmax(2)
        x = F.pad(x[:, :-n].sigmoid(), (0, n))
        softmaxLos = X[torch.arange(len(X), device=X.device)[:, None, None], child, y[:, None]].sum(2).exp_()
        childLos = (1 - x[:, child]).prod(2)
        parentLos = x[:, parent].prod(2)
        return -(softmaxLos * childLos * parentLos).sum(1).clip_(1e-40).log_().mean()
```
